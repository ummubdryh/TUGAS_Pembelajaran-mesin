# BAGIAN C- langkah Latihan

```python
import pandas as pd
data = {'luas':[36,70,45,90,54], 'kamar':[2,3,2,4,3],
        'lokasi':['Pinggir','Tengah','Pinggir','Tengah','Pinggir'],
        'harga':[250,600,320,800,400]}  # LABEL
df = pd.DataFrame(data)
X = df[['luas','kamar','lokasi']]   # fitur
y = df['harga']

# Label numerik kontinu -> REGRESI
print(df['harga'].tolist())
# Label kategori -> KLASIFIKASI
df['kategori'] = ['Murah','Mahal','Murah','Mahal','Murah']

from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test = train_test_split(
    X, y, test_size=0.4, random_state=42)
print('Latih:', len(X_train), ' Uji:', len(X_test))
```

# BAGIAN E- refleksi

# Apa perbedaan supervised, unsupervised, dan reinforcement learning?


supervised pembelajaran yang menggunakan data yang sudah ada label. lebih mudahnya belajar dengan kunci jawaban

unsupervised pembelajaran yang menggunakan data yang tidak memiliki label. gunanya untuk menemukan struktur yang tersembunyi

reinforcement pembelajaran berdasarkan rewand atau punishment, jadi dia belajar dari tindakan yang kita kasih (reward atau punisment)

# Mengapa pembagian data latih dan uji penting?


karena tujuan kita pengen tahu apakah model benar-benar bisa melakukan prediksi pada data baru atau belajar dari data yang digunakan saat latihan.

 data latih digunakan untuk melatih model sedangkan data uji itu digunakan buat melihat seberapa baik model bekerja pada data yang memang belum pernah di pelajari.

 jadi dengan adanya data uji ini kita bisa tahu apakah model sudah cukup baik atau masih perlu di perbaiki

# Mengapa alur kerja ML disebut iteratif?

karena alur kerja ML ini prosesnya dilakukan berulang, supaya kita mendapatkan model yang baik.

jadi setelah model di buat, kita melakukan pengujian dan evaluasi kalau masih kurang baik maka model diperbaiki kemudian di latih dan di uji kembali
